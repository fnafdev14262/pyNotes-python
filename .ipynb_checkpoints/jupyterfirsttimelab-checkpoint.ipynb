```python
def BasicCalc():
    num1 = float(input("First Number: "))
    op = input("What's the operator: ")
    num2 = float(input("Second Number: "))
    
    if op == "+":
        print(num1 + num2)
    elif op == "-":
        print(num1 - num2) 
    elif op == "%":
         if num2 == 0:
            print("Error: Cannot calculate remainder with zero!")
         else:
             print(num1 % num2)
    elif op == "/":
        if num2 == 0:
            print("Error: You cannot divide by zero!")
        else:
            print(num1 / num2)
    elif op == "*":
        print(num1 * num2)
    elif op == "**":
        if num1 > 100 and num2 > 100:
            print("Error: Those numbers are too big!")
        else:
            print(num1 ** num2)
    else:
        print("Error 404! Please use + - * / **")
```

```python
import random

def Reverser():
    # --- Tactical Header ---
    print("\n" + "="*40)
    print("       ARCHIVUM WORD-COGITATOR       ")
    print("              V5.1.0      ")
    print("="*40)

    text = input("\n[VOX-INPUT] Enter word or sentence: ")
    
    if not text.strip():
        print("\n[!] ERROR: VOID INPUT DETECTED. REBOOT RITUAL.")
        return

    lower = text.lower()
    reversed_text = text[::-1]
    rev_lower = reversed_text.lower()

    # --- Data Extraction ---
    print("\n--- ---")
    print(f"Lowercase: {lower}")
    print(f"Reversed:  {reversed_text}")
    print(f"Rev-Lower: {rev_lower}")

    # --- Word & Sentence Logistics ---
    words = text.split()
    sentences = sum(1 for char in text if char in ".!?")
    if sentences == 0 and len(words) > 0: sentences = 1
    
    print(f"\n--- [ LOGISTICS ] ---")
    print(f"Word count:     {len(words)}")
    print(f"Sentence count: {sentences}")

    # --- Character Counters ---
    vowels = sum(1 for char in lower if char in "aeiou")
    conson = sum(1 for char in lower if char in "bcdfghjklmnpqrstvwxyz")
    special = len(text) - (vowels + conson)

    print(f"\n--- [ SCAN: CHARACTER DATA ] ---")
    print(f"Vowels:      {vowels}")
    print(f"Consonants:  {conson}")
    print(f"Symbols/Num: {special}")

    # --- The Scrambler ---
    char_list = list(text)
    random.shuffle(char_list)
    print(f"\nScrambled: {''.join(char_list)}")

    # --- The Vault (Encryption & Corruption) ---
    print("\n--- [ THE VAULT: ENCRYPTION ] ---")
    glitch_chars = "!@#$%^&*"
    glitched = "".join(c + random.choice(glitch_chars) if random.random() > 0.7 else c for c in text)
    print(f"Glitched:  {glitched}")

    encrypted = ""
    for char in text:
        if char.isalpha():
            start = ord('a') if char.islower() else ord('A')
            encrypted += chr((ord(char) - start + 3) % 26 + start)
        else:
            encrypted += char
    print(f"Encrypted (Shift 3): {encrypted}")

    # --- Intelligence Report ---
    print("\n--- [ INTELLIGENCE REPORT ] ---")
    spooky = ["dark", "ghost", "night", "freddy", "scary", "lost"]
    secret_names = ["afton", "bonnie", "chica", "foxy"]
    
    mood = "SPOOKY VIBES DETECTED" if any(s in lower for s in spooky) else "Neutral/Safe"
    found = [n for n in secret_names if n in lower]

    print(f"Mood: {mood}")
    print(f"Hidden Files: {', '.join(found) if found else 'None'}")

    # --- Logic Analysis ---
    print("\n--- [ LOGIC ANALYSIS ] ---")
    if lower == rev_lower and len(lower) > 1:
        print("[!] PALINDROME DETECTED")
    
    if len(text) >= 8 and special > 0:
        print("Password Strength: STRONG")
    elif len(text) >= 8:
        print("Password Strength: WEAK (Needs Symbols)")
    else:
        print("Password Strength: TOO SHORT")

# --- MASTER RITUAL LOOP ---
if __name__ == "__main__":
    while True:
        Reverser()
        choice = input("\n[VOX] Review more files? (y/n): ").lower()
        if choice != 'y':
            print("\n" + "="*40)
            print("EXITING ARCHIVUM... AVE OMNISSIAH       ")
            print("="*40)
            print("Lead Developer: Fnafdev")
            print("Software Name: ARCHIVUM WORD-COGITATOR")
            print("Version: 5.1.0")
            break
```

```python
import random

def GuessGame():
    secret = random.randint(1, 100)
    guess = int((input("Guess the number")))
    tries = 0

                 
    # --- The Logic --- #
    if guess == secret:
       print("You Won!")
    else:
        while guess != secret:
             # Everything indented here keeps repeating
             guess = int(input("Try again: "))
             tries += 1
             if guess < secret:
                print("Too low! ")
             elif guess > secret:
                  print("Too high! ")
             elif guess == secret:
                  print("You Won! ")
                  input("Continue (press space)")
                  print("You took " + str(tries) + " guesses!")
                  print("Quitting")
                  input("Continue (press space)")
                  print("Closing game...")
                  return  # This exits the function immediately!
            
```

```python
def AmmoScan():
    # =====================================================================
    # ULTRAMARINE AMMO LOGISTICS - BEGINNER EDITION (V1.0.0)
    # =====================================================================

    # --- THE HEADER ---
    print("==============================")
    print("   BOLTER AMMO SCAN ACTIVE    ")
    print("==============================")

    # Input Collection
    user_data = input("Enter all ammo counts (separated by spaces): ")
    ammo_counts = [int(x) for x in user_data.split()]

    # Initializing Counters
    crit = 0
    low = 0
    optimal = 0

    # --- STEP 2: THE LOOP ---
    for rounds in ammo_counts:
        # --- STEP 3: THE LOGIC ---
        if rounds < 10:
            status = "!!! CRITICAL !!!"
            crit += 1
        elif rounds < 50:
            status = "LOW"
            low += 1
        else:
            status = "OPTIMAL"
            optimal += 1
        
        # --- STEP 4: THE OUTPUT ---
        print(f">> Unit has {rounds} rounds, and the units status is: {status}.")

    # --- BATTLE SUMMARY ---
    print("=========================================================")
    print("                   BATTLE SUMMARY                        ")
    print("=========================================================")
    print(f">> CRITICAL UNITS: {crit}")
    print(f">> LOW UNITS: {low}")
    print(f">> OPTIMAL UNITS: {optimal}")
    print(f">> TOTAL SCANNED: {len(ammo_counts)}")
    
        # --- CRITICAL ALERT LOGIC ---
    total = len(ammo_counts)
    # Check if more than 50% are critical
    is_danger = crit > (total / 2)
    print("========================")
    print("SQUAD STRENGTH CHECK")
    print("========================")
    if is_danger:
        print("[!!] ALERT [!!]")
        print("[!!] RETURN TO MACRAGGE [!!}")
    else:
        print(">> SQUAD STRENGTH FINE ")  

    # --- THE EXIT ---
    print("==============================")
    print("       SCAN COMPLETE.         ")
    print("      COURAGE AND HONOR.      ")
    print("==============================")

```
