```python
def numberKwarg(**number):
    # Fixed the syntax error by adding a key name inside the brackets
    # and added a space after 'is' for better readability
    print('my number is ' + number['value'])
    
    # Alternatively, if you want to print all keyword arguments:
    # print('my number is', number)
```


Note: When using `**number`, it collects all keyword arguments into a dictionary. You need to specify which key you want to access. If you call this function with `numberKwarg(value="42")`, then `number['value']` would contain "42".

```python
numberKwarg(value="42")
```

```python
def calc(a, op, b):
    if op == '+':
      return a + b
    if op == '-':
      return a - b
    if op == '*':
      return a * b
    if op == '%':
      return a % b
    else:
         print:("invalid")
```
