```python
def calc(a, op, b):
    try:
        if op == '+':
            return a + b
        elif op == '-':
            return a - b
        elif op == '*':
            return a * b
        elif op == '%':
            return a % b
        elif op == '**':
            # Pre-check: Limit exponent size to prevent hanging/MemoryError
            if b > 10000 and a > 1:
                return "Result too large to calculate safely"
            return a ** b
        else:
            return "invalid operator"
            
    except OverflowError:
        return "Error: Number too large for floating-point"
    except MemoryError:
        return "Error: Result exceeded system memory"
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"

```

```python
print("calc(3 '**' 5)")
```
