```python
import random
import string
```

```python
def generate_password(length=12):
    # Combine lowercase, uppercase, digits, and punctuation
    characters = string.ascii_letters + string.digits + string.punctuation
    
    # Generate password by choosing random characters
    password = ''.join(random.choice(characters) for i in range(length))
    return password

# Get user input for length
length = int(input("Enter password length: "))
print("Generated Password:", generate_password(length))
```

```python
def switch(text: str) -> str:
    """
    Reverses the input text and joins it with the original text.
    
    Args:
        text (str): The input text to be processed.
        
    Returns:
        str: The original text joined with its reversed version.
    """
    reversed_text = ''.join(reversed(text))
    return text + reversed_text
```
